```python
# Diccionario para almacenar los datos de los clientes
portafolio = {}

# Función para ingresar un cliente
def ingresar_cliente():
    clear_output()  # Limpia la salida

    # Widgets para ingresar datos
    cliente = widgets.Text(description='Cliente', placeholder='Ingresa el Nombre')
    monto_invertido = widgets.FloatText(description='Monto Invertido', value=0.0)
    rendimiento_anual = widgets.FloatText(description='Rendimiento Anual (%)', value=0.0)

    # Botón para guardar cliente
    button = widgets.Button(description="Guardar")
    output = widgets.Output()

    def on_button_clicked(b):
        # Guardar los datos en el diccionario
        if cliente.value:
            portafolio[cliente.value] = {
                "Monto Invertido": monto_invertido.value,
                "Rendimiento Anual": rendimiento_anual.value
            }
            with output:
                clear_output()
                print(f"Cliente {cliente.value} guardado en el portafolio.")
        else:
            with output:
                clear_output()
                print("Por favor, ingresa un nombre de cliente.")

    button.on_click(on_button_clicked)

    display(cliente, monto_invertido, rendimiento_anual, button, output)

# Ejecutar la función para ingresar cliente
ingresar_cliente()
```

# ___Valor Presente Neto (VPN)___

Es un método de valoración de inversiones que se utiliza para determinar la rentabilidad de un proyecto o inversión. Se calcula como la diferencia entre el valor presente de los flujos de caja futuros y el costo inicial de la inversión

### **VPN de cada flujo de caja**

```python
df.columns = df.columns.str.replace('-','_') # Cambia los - por guión bajo (_)
df.columns = df.columns.str.replace('A\x96o_', 'Año_', regex=False) # Cambia la cadena A\x96o_ por la palabra Año

print(df.columns.tolist())
```

```python
# Tasa de descuento

Tasa_de_descuento = 0.20

# Calcular el VPN

def calcular_VPN(df, Tasa_de_descuento):
  VPN = []
  proyecto = []
  for proyecto in df.index:
    # Excluir la primera columna (nombre del proyecto) al convertir a float

      Flujos_de_caja = df.loc[proyecto].values[1:].astype(float)
      Valor_VPN = npf.npv(Tasa_de_descuento, Flujos_de_caja)
      VPN.append(round(Valor_VPN, 2))

    # Agregar columna de DataFrame para asignar el cálculo del valor de la VPN

  df['VPN'] = VPN
  return df
```

```python
df = calcular_VPN(df, Tasa_de_descuento)
print("El Valor Presente Neto de cada proyecto es de:")
print(df[['Proyecto', 'VPN']])
```

```python
# Definir el nombre de las columnas

columnas_para_TIR = ['Inversion_Inicial', 'Año_1', 'Año_2', 'Año_3', 'Año_4', 'Año_5', 'Año_6', 'Año_7', 'Año_8', 'Año_9', 'Año_10']

# Calcular la TIR

def calcular_el_rendimiento(columnas_para_TIR, Tasa_de_descuento):
    TIR = []
    for proyecto in df.index:
        # Accessing the columns using the updated names in 'columnas_para_TIR'
        Flujos_de_caja = df.loc[proyecto][columnas_para_TIR].astype(float)
        Valor_TIR = npf.irr(Flujos_de_caja)
        TIR.append(round(Valor_TIR, 2))

    # Agregar columna de DataFrame para asignar el cálculo del valor de la TIR

    df['TIR'] = TIR
    return df
```

```python
df = calcular_el_rendimiento(columnas_para_TIR, Tasa_de_descuento)
print("El rendimiento de cada proyecto es de:")
print(df[['Proyecto', 'TIR']])
```

# ___¿Cómo se relacionan el VPN y la TIR con la toma de decisiones sobre proyectos de inversión?___

El VPN mide la rentabilidad actual de un proyecto, considerando el valor temporal del dinero.
Se prefieren proyectos con VPN positivo, indicando ganancias por encima de la inversión inicial.

Mientras que la TIR epresenta la tasa de rendimiento esperada de un proyecto, donde el VPN se vuelve cero.
Si la TIR supera la tasa de descuento mínima, el proyecto se considera atractivo.
