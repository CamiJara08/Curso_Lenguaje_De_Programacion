# __Gestión de Portafolios Financieros__

## ___Parte 1: Declaración de Variables y Listas___

### ___Datos de la inversión___

Clientes: Juan Pérez, Maria Rodríguez, Carlos López, Ana Gómez
Montos invertidos (en dólares): 10000, 15000, 20000, 18000
Rendimientos anuales (%): 5.5, 6.0, 4.0, 7.5

### __Creación de la lista__

```python
# Creación de la lista

Cliente= ["Juan Pérez","Maria Rodríguez","Carlos López","Ana Gómez"]
print(Cliente)
monto_invertido= [10000,15000,20000,18000]
print(monto_invertido)
rendimiento_anual= [5.5,6.0,4.0,7.5]
print(rendimiento_anual)

```

### __Tuplas de información__

```python
# Tuplas

Cliente_1= ("Juan Pérez", 10000, 5.5)
print(Cliente_1)
CLiente_2= ("Maria Rodríguez", 15000, 6.0)
print(CLiente_2)
Cliente_3= ("Carlos López", 20000, 4.0)
print(Cliente_3)
Cliente_4= ("Ana Gómez", 18000, 7.5)
print(Cliente_4)
```

## ___Parte 2: Uso de Diccionarios y Condicionales___

### __Almacenar los Datos en un Diccionario__

```python
# Datos del Diccionario

portafolio = {
    "Juan Pérez": {"monto_invertido": 10000, "rendimiento_anual": 5.5},
    "Maria Rodríguez": {"monto_invertido": 15000, "rendimiento_anual": 6.0},
    "Carlos López": {"monto_invertido": 20000, "rendimiento_anual": 4.0},
    "Ana Gómez": {"monto_invertido": 18000, "rendimiento_anual": 7.5}
}

# Evaluación de los rendimientos mayor al 5%

for Cliente, datos in portafolio.items():
    if datos["rendimiento_anual"] > 5:
        print(f"{Cliente}' tiene un rendimiento anual de {datos['rendimiento_anual']}%")

    elif datos["rendimiento_anual"] < 5:
         print(f"{Cliente} tiene un rendimiento anual menor al 5%")
```

## ___Parte 3: Cálculos con ciclos___

### __Calcular el rendimiento de un portafolio__

```python
# Rendimiento del portafolio

rendimiento_total = 0
for Cliente, datos in portafolio.items():
    rendimiento_cliente = (datos["monto_invertido"] * datos["rendimiento_anual"]) / 100
    rendimiento_total += rendimiento_cliente

print(f"El rendimiento total del portafolio es: ${rendimiento_total:.2f}")
```

```python
# Mayor rendimiento del portafolio

Max_rendimiento = 0
Cliente_Max_rendimiento = ""

for Cliente, datos in portafolio.items():
    rendimiento_cliente = (datos["monto_invertido"] * datos["rendimiento_anual"]) / 100

    if rendimiento_cliente > Max_rendimiento:
        Max_rendimiento = rendimiento_cliente
        Cliente_Max_rendimiento = Cliente

print(f"El Cliente con el mayor rendimiento es {Cliente_Max_rendimiento} con un rendimiento de ${Max_rendimiento:.2f}")
```

## ___Parte 4: Modificación de Portafolio___

### __Actualizar el Monto Invertido de un Cliente__

```python
# Carlos López ha decidio aumentar su inversión a 25000 dolares

portafolio["Carlos López"]["monto_invertido"] = 25000

print("Se actualizó el monto invertido de Carlos López a 25000 dólares")

print("El portafolio actualizado quedó así:")
print("\n")

for cliente, datos in portafolio.items():
    print(f"{cliente}: monto invertido de ${datos['monto_invertido']} y un rendimiento anual del {datos['rendimiento_anual']}%")
```

```python
monto_invertido = 25000

for Cliente, datos in portafolio.items():
  if Cliente == "Carlos López":
    (datos["monto_invertido"]) = monto_invertido

    print(f"Se actualizó el monto invertido de Carlos López a ${monto_invertido}")
```

### __Añadir un nuevo cliente__
